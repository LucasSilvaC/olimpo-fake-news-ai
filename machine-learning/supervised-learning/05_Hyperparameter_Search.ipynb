# Busca de hiperparâmetros — índice

As buscas supervisionadas estão separadas em notebooks próprios:

- [06 — Grid Search](06_Grid_Search.ipynb): percorre todas as combinações discretas configuradas.
- [07 — Random Search](07_Random_Search.ipynb): amostra combinações de distribuições e valores categóricos.

Execute primeiro [01_Data_Prep.ipynb](01_Data_Prep.ipynb) para gerar os artefatos com `taxa_stop_words`. As duas buscas usam F1 macro em validação cruzada estratificada e mantêm o conjunto de teste fora da seleção. Os espaços e a quantidade de candidatos diferem; consulte a análise de sensibilidade em cada notebook ao interpretar os hiperparâmetros.
